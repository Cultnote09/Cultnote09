```python
import requests
import csv

from bs4 import BeautifulSoup
from urllib.parse import urljoin


base_url = "https://realpython.github.io/fake-jobs/"


def scrape_jobs():
    response = requests.get(base_url, timeout=10)
    response.raise_for_status()

    soup = BeautifulSoup(response.text, "html.parser")

    jobs = []

    job_cards = soup.find_all("div", class_="card-content")

    for job in job_cards:

        title_element = job.select_one("h2.title")
        company_element = job.select_one("h3.subtitle")
        location_element = job.select_one("p.location")

        card = job.parent
        link_element = card.select_one("a.card-footer-item")

        title = (
            title_element.get_text(strip=True)
            if title_element
            else None
        )

        company = (
            company_element.get_text(strip=True)
            if company_element
            else None
        )

        location = (
            location_element.get_text(strip=True)
            if location_element
            else None
        )

        if link_element and link_element.get("href"):
            link = urljoin(base_url, link_element["href"])
        else:
            link = None

        jobs.append({
            "title": title,
            "company": company,
            "location": location,
            "link": link
        })

    return jobs


def save_to_csv(jobs):
    fields = ["title", "company", "location", "link"]

    with open("jobs.csv", "w", newline="", encoding="utf-8") as output_file:
        writer = csv.DictWriter(
            output_file,
            fieldnames=fields
        )

        writer.writeheader()
        writer.writerows(jobs)


def main():
    jobs = scrape_jobs()
    save_to_csv(jobs)

    print("Jobs saved to jobs.csv")


if __name__ == "__main__":
    main()





```
